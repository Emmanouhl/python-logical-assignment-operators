**MINI PROJECT : INVENTORY MANAGEMENT SYSTEM (PRE-TESTING)**
=

```python
stock = int(input("How many items are in stock: "))
stock -= int(input("How many items were sold: "))
print("Remaining stock after sales is: ", stock)
stock += int(input("How many new items arrived: "))
print("Updated stock after restocking is: ", stock)
stock //= int(input("How many items fit into one carton: "))
print("Number of full cartons is: ", stock)
stock %= int(input("How many items remaining is lost: "))
print("Remaining loose items is: ", stock)
```

**MINI PROJECT : INVENTORY MANAGEMENT SYSTEM (MAIN PROGRAM)**
=

```python
stock = int(input("How many items are in stock: "))
stock -= int(input("How many items were sold: "))
print("Remaining stock after sales is: ", stock)
updated_stock = stock + int(input("How many new items arrived: "))
print("Updated stock after restocking is: ", updated_stock)
numbers_in_carton = updated_stock // int(input("How many items fit into one carton: "))
print("Number of full cartons is: ", numbers_in_carton)
remaining_items = updated_stock % numbers_in_carton
print("Remaining loose items is: ", remaining_items)

```
