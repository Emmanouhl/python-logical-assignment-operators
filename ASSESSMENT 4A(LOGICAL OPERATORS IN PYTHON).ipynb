*EXERCISE ONE (SECURE LOGIN SYSTEM)*

```python
username_correct = True
password_correct = True
account_locked = False

grant_access = username_correct and password_correct and not(account_locked)
if grant_access == True:
    print("Access Granted")
else:
    print("Access Denied")
```

```python
username_correct = True
password_correct = True
account_locked = True

denial_access = not (username_correct) and not (password_correct) and (account_locked)
if denial_access == False:
    print("Access Denial")
```
